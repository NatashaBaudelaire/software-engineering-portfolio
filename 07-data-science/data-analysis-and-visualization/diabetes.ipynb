1. Data Preparation and Cleaning
The Pima dataset is known for having 0 values in columns where that is biologically impossible (such as Blood Pressure or BMI). These zeros are actually missing data.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Loading the dataset (adjust the path as needed)
df = pd.read_csv('diabetes.csv')

# 1. Handling Missing Data: Replacing 0 with NaN in specific columns
cols_with_invalid_zero = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']
df[cols_with_invalid_zero] = df[cols_with_invalid_zero].replace(0, np.nan)

# Filling NaNs with the median of each column (common strategy to avoid outliers)
df.fillna(df.median(), inplace=True)

# 2. Handling Anomalies (Outliers)
# Example: Remove records where BMI is extreme (optional, depending on the analysis)
# Here we will focus on the inconsistency cleaning already performed above.

print("Data cleaned. Summary of nulls:\n", df.isnull().sum())
```

2. Measures of Central Tendency, Dispersion and Distribution
We calculate the mean, median (central tendency), standard deviation (dispersion) and quartiles.

```python
# Descriptive Statistics
statistics = df.describe().T
statistics['median'] = df.median()
print(statistics[['mean', 'median', 'std', 'min', 'max']])

# Distribution Visualization (Histograms)
df.hist(figsize=(12, 10), bins=20, color='skyblue', edgecolor='black')
plt.suptitle("Attribute Distribution - Pima Indians Diabetes")
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

3. Class Balance
It is essential to check whether there are far more "Non-Diabetic" (0) cases than "Diabetic" (1) cases, since this affects model training.

```python
# Checking the balance of the 'Outcome' column
plt.figure(figsize=(6, 4))
sns.countplot(x='Outcome', data=df, palette='viridis')
plt.title('Target Class Distribution (0: Healthy, 1: Diabetes)')
plt.show()

print("Class proportions:\n", df['Outcome'].value_counts(normalize=True))
```

4. Correlation Investigation
The correlation matrix helps us understand which variables have the strongest relationship with the diabetes diagnosis.

```python
# Pearson Correlation Calculation
plt.figure(figsize=(10, 8))
correlation_matrix = df.corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Matrix Between Attributes")
plt.show()

# Specific correlation with the Target (Outcome)
corr_with_target = correlation_matrix['Outcome'].sort_values(ascending=False)
print("Correlation of attributes with the diagnosis:\n", corr_with_target)
```

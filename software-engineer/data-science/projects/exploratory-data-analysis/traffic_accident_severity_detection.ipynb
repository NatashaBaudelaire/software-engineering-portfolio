# Exploratory Data Analysis (EDA): Traffic Accident Severity Detection


## 1. Setup and Data Loading

```sql
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import (
    LabelEncoder,
    OneHotEncoder,
    Normalizer,
    MinMaxScaler,
    StandardScaler,
    RobustScaler,
    QuantileTransformer,
    PowerTransformer
)
from sklearn.decomposition import PCA
from sklearn.impute import SimpleImputer
from sklearn.utils import resample

from scipy import stats
```

```sql
# Loading the dataset
data = '../../datasets/traffic_accident_severity_detection.csv'
data = pd.read_csv(data, sep=';', encoding='iso-8859-1')

# Initial cleanup of column names (remove extra spaces and convert to lowercase)
data.columns = [col.strip() for col in data.columns]
data.columns = [col.lower() for col in data.columns]

# Translate column names to English for readability
data = data.rename(columns={
    'sex': 'sex',
    'seat_belt': 'seat_belt',
    'intoxication': 'intoxication',
    'pedestrian': 'pedestrian',
    'passenger': 'passenger',
    'driver': 'driver',
    'old_severity_code': 'old_severity_code',
    'severity_desc': 'severity_desc',
    'age': 'age',
    'report_datetime': 'report_datetime',
    'vehicle_type': 'vehicle_type',
})

display(data.head())
print("\n\n")
display(data.tail())
print("\n\n")
display(data.sample(5))
```

## 2. Data Inspection and Cleaning
In this step, we check data types, missing values, and the consistency of the variables.


```sql
# General information about the columns
data.info()
```

```sql
# Checking categorical values

# Unique values of the categorical variables
print(data["sex"].unique())
print(data["seat_belt"].unique())
print(data["intoxication"].unique())
print(data["pedestrian"].unique())
print(data["passenger"].unique())
print(data["driver"].unique())
print("\n")


# Value counts (categorical variable)
print(data["old_severity_code"].value_counts())
print("\n\n")

# Checking for possible numeric inconsistencies
check_01 = data[(data["age"] < 0) | (data["age"] > 100)]
print(check_01)
```

```sql
# Checking for null values

display(data.isnull().sum())

# Missing data (percentage)
print("----------------Percentage------------------")
missing_percent = data.isnull().mean() * 100
print(missing_percent.sort_values(ascending=True))
print("\n\n")
```

```sql
# Basic treatment: convert report_datetime to datetime
data['report_datetime'] = pd.to_datetime(data['report_datetime'], format='%d/%m/%Y %H:%M', errors='coerce')
data['hour'] = data['report_datetime'].dt.hour
data['day_of_week'] = data['report_datetime'].dt.day_name()
```

## 3. Target Variable Analysis: Severity
The `severity_desc` variable is our main target.


```sql
# Severity distribution
plt.figure(figsize=(10, 5))
sns.countplot(data=data, x='severity_desc', palette='viridis')
plt.title('Distribution of Accident Severity')
plt.xlabel('Severity')
plt.ylabel('Frequency')
plt.xticks(rotation=45)
plt.show()
```

## 4. Profile of Those Involved
Analysis by sex, age, and role (driver/passenger).


```sql
# Distribution by sex and severity
plt.figure(figsize=(10, 6))
sns.countplot(data=data, x='sex', hue='severity_desc')
plt.title('Severity by Sex')
plt.show()
```

```sql
# Age distribution
plt.figure(figsize=(10, 6))
sns.histplot(data['age'].dropna(), bins=30, kde=True, color='skyblue')
plt.title('Age Distribution of Those Involved')
plt.show()
```

## 5. Risk and Safety Factors
Seat belt use and intoxication.


```sql
# Seat belt vs severity
plt.figure(figsize=(10, 6))
sns.countplot(data=data, x='seat_belt', hue='severity_desc')
plt.title('Impact of Seat Belt Use on Severity')
plt.show()
```

```sql
# Intoxication
print('Reported intoxication frequency:')
print(data['intoxication'].value_counts())
```

## 6. Analysis by Vehicle Type
Which vehicles are most involved in severe accidents?


```sql
# Top 10 vehicle types
top_vehicles = data['vehicle_type'].value_counts().nlargest(10).index
plt.figure(figsize=(12, 6))
sns.countplot(data=data[data['vehicle_type'].isin(top_vehicles)],
              y='vehicle_type', hue='severity_desc')
plt.title('Severity by Vehicle Type (Top 10)')
plt.show()
```

## 7. Temporal Analysis
Peak occurrence times.


```sql
# Accidents by hour of day
plt.figure(figsize=(12, 6))
sns.countplot(data=data, x='hour', color='salmon')
plt.title('Distribution of Accidents by Hour of Day')
plt.xlabel('Hour')
plt.show()
```
